# Phase 4 — Data Cleaning

## Create a Working Copy

To preserve the original dataset, all preprocessing steps will be applied to a separate copy.

This approach ensures that the raw data remains unchanged and allows us to reproduce or modify the cleaning process at any time.

```python
# Create a copy of the original dataset
clean_df = df.copy()

# Verify
clean_df.head()
```

### Handle Missing Values — Embarked

The `Embarked` feature contains only a small number of missing observations.

Since it is a categorical variable with less than 1% missing values, mode imputation is considered the most appropriate strategy.

```python
clean_df["Embarked"].isnull().sum()
```

```python
clean_df["Age"].isnull().sum()
```

```python
# Age
clean_df["Age"] = clean_df["Age"].fillna(clean_df["Age"].median())

# Embarked
clean_df["Embarked"] = clean_df["Embarked"].fillna(clean_df["Embarked"].mode()[0])
```

## Outlier Assessment

Outliers are observations that differ significantly from the rest of the data.

While some outliers may result from data entry errors, others may represent legitimate observations. Therefore, outliers should be carefully evaluated before deciding whether they should be treated or retained.

In this step, we will inspect the numerical features visually using boxplots.

```python
# Select numerical features excluding the target variable
outlier_features = ["Age", "Fare", "SibSp", "Parch"]

plt.figure(figsize=(12, 8))

for i, feature in enumerate(outlier_features, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(y=clean_df[feature])
    plt.title(feature)

plt.tight_layout()
plt.show()
```

## Optimize Data Types

Some features represent categorical information but are currently stored as integer or object data types.

Converting these features to the `category` data type improves memory efficiency and makes their semantic meaning clearer.

```python
categorical_columns = ["Survived", "Pclass", "Sex", "Embarked"]

for column in categorical_columns:
    clean_df[column] = clean_df[column].astype("category")
```

## Final Cleaning Validation

After completing all preprocessing steps, the dataset is validated to ensure that the identified data quality issues have been addressed successfully.

# Phase 5 — Exploratory Data Analysis

### Business Question

Did passenger gender influence the likelihood of survival?

```python
clean_df["Survived"] = clean_df["Survived"].astype(int)
gender_summary = (
    clean_df
    .groupby("Sex", observed=True)["Survived"]
    .agg(["count", "sum", "mean"])
    .rename(columns={
        "count": "Passengers",
        "sum": "Survived",
        "mean": "Survival Rate"
    })
)

gender_summary["Survival Rate"] = (
    gender_summary["Survival Rate"] * 100
).round(2)

gender_summary
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    data=clean_df,
    x="Sex",
    y="Survived",
    estimator="mean",
    errorbar=None
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.ylim(0,1)

plt.show()
```

## Business Question 2



Did the passenger class influence the probability of survival?



```python
pclass_summary = (
    clean_df
    .groupby("Pclass", observed=True)
    .agg(
        Passengers=("PassengerId", "count"),
        Survived=("Survived", "sum")
    )
)

pclass_summary["Survival Rate (%)"] = (
    pclass_summary["Survived"] /
    pclass_summary["Passengers"] * 100
).round(2)

pclass_summary
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    data=pclass_summary.reset_index(),
    x="Pclass",
    y="Survival Rate (%)"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")

plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

## Business Question 3


Did passenger age influence survival?


```python
age_summary = (
    clean_df
    .groupby("Survived")
    .agg(
        Passengers=("PassengerId", "count"),
        Mean_Age=("Age", "mean"),
        Median_Age=("Age", "median")
    )
    .round(2)
)

age_summary
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=clean_df,
    x="Survived",
    y="Age"
)

plt.title("Age Distribution by Survival")
plt.xlabel("Survived")
plt.ylabel("Age")

plt.tight_layout()
plt.show()
```

## Business Question 4
Did ticket fare influence survival?

```python
fare_summary = (
    clean_df
    .groupby("Survived")
    .agg(
        Mean_Fare=("Fare", "mean"),
        Median_Fare=("Fare", "median")
    )
    .round(2)
)

fare_summary
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=clean_df,
    x="Survived",
    y="Fare"
)

plt.title("Fare Distribution by Survival")
plt.xlabel("Survived")
plt.ylabel("Fare")

plt.tight_layout()
plt.show()
```

## Business Question 5

Did the embarkation port affect passenger survival?

```python
embarked_summary = (
    clean_df
    .groupby("Embarked", observed=True)
    .agg(
        Passengers=("PassengerId", "count"),
        Survived=("Survived", "sum")
    )
)

embarked_summary["Survival Rate (%)"] = (
    embarked_summary["Survived"] /
    embarked_summary["Passengers"] * 100
).round(2)

embarked_summary
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    data=embarked_summary.reset_index(),
    x="Embarked",
    y="Survival Rate (%)"
)

plt.title("Survival Rate by Embarkation Port")

plt.tight_layout()
plt.show()
```

## Business Question 6

Did family size influence passenger survival?

```python
clean_df["FamilySize"] = clean_df["SibSp"] + clean_df["Parch"] + 1
```

```python
family_summary = (
    clean_df
    .groupby("FamilySize")
    .agg(
        Passengers=("PassengerId", "count"),
        Survived=("Survived", "sum")
    )
)

family_summary["Survival Rate (%)"] = (
    family_summary["Survived"] /
    family_summary["Passengers"] * 100
).round(2)

family_summary
```

```python
plt.figure(figsize=(9,5))

sns.lineplot(
    data=family_summary.reset_index(),
    x="FamilySize",
    y="Survival Rate (%)",
    marker="o"
)

plt.title("Survival Rate by Family Size")

plt.tight_layout()
plt.show()
```

```python
clean_df["IsAlone"] = (clean_df["FamilySize"] == 1).astype(int)
```

```python
alone_summary = (
    clean_df
    .groupby("IsAlone")
    .agg(
        Passengers=("PassengerId", "count"),
        Survived=("Survived", "sum")
    )
)

alone_summary["Survival Rate (%)"] = (
    alone_summary["Survived"] /
    alone_summary["Passengers"] * 100
).round(2)

alone_summary
```

```python
plt.figure(figsize=(6,5))

sns.barplot(
    data=alone_summary.reset_index(),
    x="IsAlone",
    y="Survival Rate (%)"
)

plt.xticks([0,1],["Not Alone","Alone"])

plt.title("Survival Rate by Traveling Status")

plt.tight_layout()
plt.show()
```

# Phase 6 — Feature Engineering

In this phase, new features are created from the existing variables to improve data representation and provide additional information for future machine learning models.

## Cabin Availability

```python
clean_df["HasCabin"] = clean_df["Cabin"].notna().astype(int)
```

## Passenger Title

```python
clean_df["Title"] = (
    clean_df["Name"]
    .str.extract(r",\s*([^\.]+)\.")
)
```

```python
clean_df["Title"] = clean_df["Title"].replace({
    "Mlle":"Miss",
    "Ms":"Miss",
    "Mme":"Mrs",
    "Lady":"Rare",
    "Countess":"Rare",
    "Capt":"Rare",
    "Col":"Rare",
    "Don":"Rare",
    "Dr":"Rare",
    "Major":"Rare",
    "Rev":"Rare",
    "Sir":"Rare",
    "Jonkheer":"Rare",
    "Dona":"Rare"
})
clean_df["Title"].value_counts()
```

## Drop Unnecessary Features

```python
model_df = clean_df.drop(
    columns=[
        "PassengerId",
        "Name",
        "Ticket",
        "Cabin"
    ]
)
```
