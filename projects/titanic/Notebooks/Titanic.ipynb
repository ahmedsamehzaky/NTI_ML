````python
import pandas as pd
titanic = pd.read_csv("data/titanic.csv")
titanic.head()
````

A quick note on this version of the dataset: unlike a lot of trimmed-down Titanic samples
you'll see online, this one has the full original columns, including a genuine unique key —
`PassengerId` — plus `Name`, `Ticket`, and `Cabin` in their raw form. A few things that changes
for us:
- **Duplicates become a much more meaningful check.** With a real unique ID per row, any
  `duplicated()` result (checked across all columns) tells you something is *actually* wrong with
  the data — it's not just "these passengers happen to share every visible feature" the way it
  would be without an ID.
- **`Cabin` is mostly missing** (most passengers, especially in lower classes, have no recorded
  cabin) — a different, more extreme missing-data situation than anything in our Employees table.
- **`Name` is real, raw text** — including each passenger's title (`Mr.`, `Mrs.`, `Miss.`,
  `Master.`, and a few rarer ones) embedded right in the string. That's genuinely useful data,
  not just a label, once you pull it apart with `.str` methods.

````python
# Task 1: Explore

# 1. Check the dimensions of the dataset (rows and columns)
print("Dataset Shape:", titanic.shape)

# 2. Inspect data types and spot missing values
print("\n--- Data Info ---")
titanic.info()

# 3. Generate summary statistics for numerical columns
print("\n--- Summary Statistics ---")
display(titanic.describe())
````

````python
# Task 2: Select

# Select specific columns into a new DataFrame to focus on key features
titanic_subset = titanic[['Survived', 'Pclass', 'Sex', 'Age', 'Fare']].copy()

# Display the first few rows to verify the selection
display(titanic_subset.head())
````

````python
# Task 3: Filter

# Apply multiple conditions: Female AND 1st Class AND Survived
female_first_class_survivors = titanic[(titanic['Sex'] == 'female') & 
                                       (titanic['Pclass'] == 1) & 
                                       (titanic['Survived'] == 1)]

print(f"Number of 1st class female survivors: {female_first_class_survivors.shape[0]}")
display(female_first_class_survivors.head())
````

````python
# Task 4: Sort

# Sort the dataset by 'Fare' in descending order (highest first)
top_10_fares = titanic.sort_values(by='Fare', ascending=False).head(10)

# Display just the names and fares to see who paid the most
display(top_10_fares[['Name', 'Fare', 'Pclass']])
````

````python
# Task 5: Create AgeGroup

# We will use pd.cut to bucket the ages into predefined categories.
bins = [0, 12.99, 19.99, 59.99, 120]
labels = ['Child', 'Teen', 'Adult', 'Senior']

# Create the categorical column
titanic['AgeGroup'] = pd.cut(titanic['Age'], bins=bins, labels=labels)

# Handle missing ages sensibly by adding an 'Unknown' category
titanic['AgeGroup'] = titanic['AgeGroup'].cat.add_categories('Unknown').fillna('Unknown')

# Check the distribution of our new feature
print(titanic['AgeGroup'].value_counts())
````

````python
# Task 6: Handle missing values

# Check initial missing values
print("Missing before cleaning:\n", titanic[['Age', 'Cabin', 'Embarked']].isnull().sum())

# Strategy 1: 'Embarked' only has 2 missing values. Impute with the most frequent port (mode).
titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])

# Strategy 2: 'Age' Impute with the median to avoid skewness from outliers.
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())
 
# fill it with 'Unknown' because the fact that a cabin is unrecorded might itself be useful data.
titanic['Cabin'] = titanic['Cabin'].fillna('Unknown')

# Re-check to ensure no missing values remain in these specific columns
print("\nMissing after cleaning:\n", titanic[['Age', 'Cabin', 'Embarked']].isnull().sum())
````

````python
# Task 7: Remove duplicates

# Count exact row duplicates across all columns
duplicate_count = titanic.duplicated().sum()
print(f"Total duplicate rows: {duplicate_count}")
````

````python
# Task 8: Group and aggregate

# Calculate the survival rate (mean of 0s and 1s) grouped by Sex
survival_by_sex = titanic.groupby('Sex')['Survived'].mean()
print("--- Survival Rate by Sex ---")
print(survival_by_sex)

# Calculate the survival rate grouped by Passenger Class
survival_by_class = titanic.groupby('Pclass')['Survived'].mean()
print("\n--- Survival Rate by Pclass ---")
print(survival_by_class)
````

````python
# Task 9: Pivot table
pivot_survival = titanic.pivot_table(index='Pclass', 
                                     columns='Sex', 
                                     values='Survived', 
                                     aggfunc='mean')

display(pivot_survival)
````

````python
# Task 10: Transform -- extract Title from Name (Simplified)

# Step 1: Split by comma and take the 2nd part -> " Mr. Owen Harris"
# Step 2: Split by dot and take the 1st part -> " Mr"
# Step 3: Strip spaces -> "Mr"
titanic['Title'] = titanic['Name'].str.split(',').str[1].str.split('.').str[0].str.strip()

# Check the unique titles extracted
print("Unique Titles:", titanic['Title'].unique())

# Group by the new Title column to check survival rates and count of passengers
survival_by_title = titanic.groupby('Title')['Survived'].agg(['mean', 'count']).sort_values(by='count', ascending=False)

print("\n--- Survival Rate by Title ---")
display(survival_by_title.head(10))
````
