## 1. Exploratory Data Analysis (EDA)

In this section, we will explore the Titanic dataset to understand its structure, assess data quality, and uncover initial patterns. Our main goal is to identify which features significantly influenced the passengers' chances of survival.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../titanic-data/titanic.csv")
```

### 1.1 Data Overview & Summary Statistics

First, we load the dataset and inspect the initial rows. We will also check the data types, identify any missing values, and generate descriptive statistics for the numerical columns to understand the data distribution.

```python
display(df.head())
df.info()
print(df.isnull().sum())
display(df.describe())
```

### 1.2 Visualizing the Target Variable (Survived)

We begin our visual analysis by looking at the target variable: `Survived`. This bar chart displays the overall count of passengers who survived (1) versus those who did not (0), giving us a clear baseline for our dataset.

```python

df['Survived'].value_counts().plot(kind='bar', color=['salmon', 'lightgreen'])
plt.title('Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

### 1.3 Age Distribution of Passengers

Next, we examine the age demographics of the people on board. This histogram illustrates the distribution of ages, helping us identify the most common age groups among the passengers.

```python
df['Age'].plot(kind='hist', bins=20, edgecolor='black', color='skyblue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
survival_by_sex = pd.crosstab(df['Sex'], df['Survived'])
survival_by_sex.plot(kind='bar', color=['salmon', 'lightgreen'])
plt.title('Survival by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.legend(['Not Survived', 'Survived'])
plt.show()
```
